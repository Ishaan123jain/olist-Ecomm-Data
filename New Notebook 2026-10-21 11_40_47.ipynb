```python
payments_path = (
    "abfss://olistcontainer@oliststorageishaan.dfs.core.windows.net/"
    "bronze/`olist_order_payments`.txt"
)

payments_df = (
    spark.read.format("csv")
    .option("header", "true")
    .option("inferSchema", "true")
    .option("quote", '"')
    .option("escape", '"')
    .load(payments_path)
)

display(payments_df.limit(10))
```

```python
from pyspark.sql import functions as F

raw_df = spark.read.text(payments_path)

payments_df = (
    raw_df
    .filter(~F.col("value").startswith("order_id,"))
    .select(
        F.split(F.col("value"), ",").alias("cols")
    )
    .select(
        F.regexp_replace(F.col("cols")[0], r'^"+|"+$', "").alias("order_id"),
        F.col("cols")[1].cast("int").alias("payment_sequential"),
        F.regexp_replace(F.col("cols")[2], r'^"+|"+$', "").alias("payment_type"),
        F.col("cols")[3].cast("int").alias("payment_installments"),
        F.regexp_replace(F.col("cols")[4], r'^"+|"+$', "").cast("double").alias("payment_value")
    )
)

display(payments_df.limit(10))
```

```python
import pandas as pd
collection=mydatabase['product_category_name_translation']
mongo_data=pd.DataFrame(list(collection.find()))
mongo_data.head()
mongo_data.drop("_id", axis=1, inplace=True)

translation_df = spark.createDataFrame(mongo_data)
```

```python
def clean_dataframe(df,name):
    print("cleaning : ",name)
    return df.dropDuplicates().na.drop("all")

orders_df = clean_dataframe(orders_df,"orders")
orders_df.show(3)
```

```python
from pyspark.sql.functions import *
orders_df = orders_df.withColumn("order_purchase_timestamp", to_date(col("order_purchase_timestamp")))\
.withColumn("order_approved_at", to_date(col("order_approved_at")))\
    .withColumn("order_delivered_carrier_date", to_date(col("order_delivered_carrier_date")))\
        .withColumn("order_delivered_customer_date", to_date(col("order_delivered_customer_date")))\
            .withColumn("order_estimated_delivery_date", to_date(col("order_estimated_delivery_date")))
orders_df.printSchema()
```

```python
# 1. Orders + Payments
orders_payments_df = orders_df.join(
    payments_df, on="order_id", how="left"
)

# 2. Add Customers
orders_customer_df = orders_payments_df.join(
    customers_df, on="customer_id", how="left"
)

# 3. Add Order Items
orders_items_df = orders_customer_df.join(
    items_df, on="order_id", how="left"
)

# 4. Add Products
orders_products_df = orders_items_df.join(
    products_df, on="product_id", how="left"
)

# 5. Add Sellers
final_df = orders_products_df.join(
    sellers_df, on="seller_id", how="left"
)

# Display results
final_df.show(3, truncate=False)
```

```python
final_df=final_df.join(translation_df, final_df.product_category_name == translation_df.product_category_name, "left")

final_df.display()
```
